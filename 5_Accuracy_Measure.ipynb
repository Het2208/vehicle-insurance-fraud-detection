```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("insurance_fraud_data_preprocessed.csv")
df.sample()  
```

```python
X = df.drop(columns = ["fraud_reported"])
y = df.iloc[: ,-1]
print(X.shape)
print(y.shape)
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.25 , random_state = 42)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

# Accuracy matrics

```python
from sklearn.metrics import accuracy_score , confusion_matrix , f1_score , precision_score , recall_score
```

## Training Accuracy

```python
y_pred_train = model.predict(X_train)
```

```python
print(f"Accuracy score : " , accuracy_score(y_train , y_pred_train))
```

```python
print("Confusion matrix : ")
confusion_matrix(y_train , y_pred_train)
```

```python
print("Precision score : ")
precision_score(y_train , y_pred_train)
```

```python
print("recall score : ")
recall_score(y_train , y_pred_train)
```

```python
print("f1 score : ")
f1_score(y_train , y_pred_train)
```

## Testing Accuracy

```python
y_pred_test = model.predict(X_test)
```

```python
print(f"Accuracy score : " , accuracy_score(y_test , y_pred_test))
```

```python
print("Confusion matrix : ")
confusion_matrix(y_test , y_pred_test)
```

```python
print("Precision score : ")
precision_score(y_test , y_pred_test)
```

```python
print("recall score : ")
recall_score(y_test , y_pred_test)
```

```python
print("f1 score : ")
f1_score(y_test , y_pred_test)
```

# Cross Validation score for 10 times 

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(model,X_train,y_train,cv=10,scoring='accuracy')
```

```python
scores.mean()
```

# Hyperparameter tunning 

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': [2, 4, 6, 8, 10, None],
    'criterion': ['gini', 'entropy'],
    'max_features': [None, 'sqrt', 'log2'],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10]
}

t = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    scoring='accuracy',
    n_jobs=-1
)

t.fit(X_train, y_train)
```

```python
t.best_score_
```

```python
t.best_params_
```

### Grid search's cross validation score 

```python
scores = cross_val_score(t,X_train,y_train,cv=10,scoring='accuracy')
```

```python
scores.mean()
```
