```python
import numpy as np
import pandas as pd

from sklearn.model_selection import (
    StratifiedKFold,
    GridSearchCV
)

from sklearn.base import clone

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)
```

```python
df = pd.read_csv("insurance_fraud_data_preprocessed.csv")

X = df.drop(columns=["fraud_reported"])
y = df["fraud_reported"].astype(int)

print(X.shape)
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
base_models = {

    "logistic": (
        Pipeline([
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                max_iter=2000,
                random_state=42
            ))
        ]),

        {
            "model__C": [0.01, 0.1, 1, 10],
            "model__solver": ["lbfgs", "liblinear"]
        }
    ),

    "decision_tree": (
        DecisionTreeClassifier(
            random_state=42
        ),

        {
            "max_depth": [2, 3, 4, 5, 7, 10],
            "min_samples_split": [2, 5, 10, 20],
            "min_samples_leaf": [1, 2, 5, 10],
            "criterion": ["gini", "entropy"],
            "max_features": [None, "sqrt", "log2"]
        }
    ),

    "random_forest": (
        RandomForestClassifier(
            random_state=42,
            n_jobs=-1
        ),

        {
            "n_estimators": [200, 300],
            "max_depth": [None, 5, 10],
            "min_samples_split": [2, 5],
            "min_samples_leaf": [1, 2, 5],
            "max_features": ["sqrt", "log2"]
        }
    ),

    "extra_trees": (
        ExtraTreesClassifier(
            random_state=42,
            n_jobs=-1
        ),

        {
            "n_estimators": [200, 300],
            "max_depth": [None, 5, 10],
            "min_samples_split": [2, 5],
            "min_samples_leaf": [1, 2, 5],
            "max_features": ["sqrt", "log2"]
        }
    ),

    "gradient_boosting": (
        GradientBoostingClassifier(
            random_state=42
        ),

        {
            "n_estimators": [100, 200],
            "learning_rate": [0.03, 0.05, 0.1],
            "max_depth": [2, 3, 4],
            "min_samples_split": [2, 5],
            "min_samples_leaf": [1, 3, 5]
        }
    ),

    "adaboost": (
        AdaBoostClassifier(
            random_state=42
        ),

        {
            "n_estimators": [100, 200, 300],
            "learning_rate": [0.03, 0.05, 0.1, 0.5, 1.0]
        }
    )
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

tuned_models = {}

for name, (model, param_grid) in base_models.items():

    print("\n" + "=" * 60)
    print("TUNING:", name)
    print("=" * 60)

    grid = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
        verbose=1
    )

    grid.fit(X_train, y_train)

    print("\nBest parameters:")
    print(grid.best_params_)

    print("Best CV ROC-AUC:")
    print(grid.best_score_)

    tuned_models[name] = grid.best_estimator_
```

```python
def generate_oof_predictions(
    X_train,
    y_train,
    X_test,
    tuned_models,
    n_splits=5
):

    skf = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=42
    )

    n_models = len(tuned_models)

    # OOF predictions
    oof = np.zeros(
        (len(X_train), n_models)
    )

    # Test predictions
    test_predictions = np.zeros(
        (len(X_test), n_models)
    )

    model_names = list(tuned_models.keys())

    for model_idx, model_name in enumerate(model_names):

        print("\n" + "=" * 60)
        print("OOF MODEL:", model_name)
        print("=" * 60)

        model = tuned_models[model_name]

        # Test predictions for every fold
        fold_test_predictions = np.zeros(
            (len(X_test), n_splits)
        )

        for fold, (train_idx, valid_idx) in enumerate(
            skf.split(X_train, y_train)
        ):

            print(
                f"Fold {fold + 1}/{n_splits}"
            )

            X_fold_train = X_train.iloc[train_idx]
            X_fold_valid = X_train.iloc[valid_idx]

            y_fold_train = y_train.iloc[train_idx]
            y_fold_valid = y_train.iloc[valid_idx]

            # Fresh copy
            fold_model = clone(model)

            # Train
            fold_model.fit(
                X_fold_train,
                y_fold_train
            )

            # OOF prediction
            valid_prob = fold_model.predict_proba(
                X_fold_valid
            )[:, 1]

            oof[
                valid_idx,
                model_idx
            ] = valid_prob

            # Test prediction
            fold_test_predictions[
                :,
                fold
            ] = fold_model.predict_proba(
                X_test
            )[:, 1]

        # Average test predictions
        test_predictions[
            :,
            model_idx
        ] = fold_test_predictions.mean(axis=1)

    return oof, test_predictions
```

```python
oof_train, stacked_test = generate_oof_predictions(
    X_train,
    y_train,
    X_test,
    tuned_models,
    n_splits=5
)
```
